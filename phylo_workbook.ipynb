# How similar is SARS-Cov-2 to other viruses?

Evolution is the process by which different living organisms are believed to have developed from earlier forms during the earth's history. The main component that evolves through time is the DNA which is passed on to the next generation. Genetic mutations, variation, and recombination are processes that modify DNA and cause variation in the population. The main drivers of evolution are natural selection and genetic drift that act on this variation. Through time, some characteristics become more common or rare in the population resulting in a change in heritable ones arising over successive generations.

Even though we do not consider viruses to be *living* organisms, they are still subject to evolution. Their genetic material undergoes the same transformations and selection, resulting in new variants and types of viruses.

Comparing the characteristics of different organisms allows us to estimate their evolutionary distance. We use genetic material as a proxy for these characteristics. We estimate evolutionary distance as the number of transformation events that would occur for one organism to become genetically identical. In practice, this means aligning genetic sequences and observing the number of transformation events. Performing multiple sequence alignment and calculating pairwise distances gives us a distance matrix.

One way of presenting the evolutionary distance between organisms is with a phylogenetic tree. Similar organisms are connected via their common ancestor, even though we do not have genetic data to prove their existence. However, these common ancestors likely existed in the past due to evolution. The most common algorithms for building phylogenetic trees are the UPGMA and Neighbor-Joining algorithms. Many frequentist and Bayesian variations of these are also available online.

````python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from Bio import SeqIO
from helper_functions import neighbor_joining, plot_nj_tree, reroot_tree, sort_children_by_leaves

# Question a

# Read the multiple sequence alignment
records = list(SeqIO.parse("data/clustal_MSA_coronavirus.fasta", "fasta"))
labels = [rec.id for rec in records]

# Encode the alignment as a 2D NumPy array:
# rows represent sequences; columns represent aligned positions
msa = np.array([np.frombuffer(str(rec.seq).upper().encode(), dtype=np.uint8) for rec in records])
print("alignment shape:", msa.shape)

# Compute the pairwise Hamming distance matrix by counting unequal
# characters at aligned positions. A gap versus a base is a mismatch;
# two gaps are a match. Distances are counts, not proportions
distances = (msa[:, None, :] != msa[None, :, :]).sum(axis=2).astype(float)

# Build the tree with our neighbour-joining implementation
tree = neighbor_joining(distances, labels)


def short_name(label):
    """Show only the virus name on the plot, without the accession."""
    return label.split("|")[0]


# Plot the tree
fig, ax = plt.subplots(figsize=(10, 10))
plot_nj_tree(tree, ax=ax, fontsize=9, label_func=short_name)
ax.set_title("a) Neighbour-joining tree of coronaviruses (Hamming distance)")
ax.set_xlabel("Hamming distance (mismatched alignment positions)")
ax.set_yticks([])
ax.spines["left"].set_visible(False)
plt.show()


````

````python
# b) Reroot the tree with Breda virus as the outgroup

def find_node(node, name):
    """Search the tree (depth-first) and return the node with the given name."""
    stack = [node]
    while stack:
        current = stack.pop()
        if current is None:
            continue
        if current.name == name:
            return current
        stack += [current.left, current.right]
    return None


# 1. Fetch the outgroup node from the tree （"Breda|NC_007447.1")
outgroup_label = next(label for label in labels if label.startswith("Breda|"))
outgroup = find_node(tree, outgroup_label)

# 2. Reroot: the new root sits halfway along the branch leading to Breda.
#    Our Node class has no extra relational attributes, so reroot_tree needs no changes.
rooted_tree = reroot_tree(tree, outgroup)
print("children of the new root:", rooted_tree.left.name, "and", rooted_tree.right.name)

# 3. Plot the rooted tree
fig, ax = plt.subplots(figsize=(10, 10))
plot_nj_tree(rooted_tree, ax=ax, fontsize=9, label_func=short_name)
ax.set_title("b) Tree rooted on Breda virus (outgroup)")
ax.set_xlabel("Hamming distance from root (mismatched alignment positions)")
ax.set_yticks([])
ax.spines["left"].set_visible(False)
plt.show()
````

````python
# c) Sort children: the left child is the subtree with fewer leaves

# Sorts the rooted tree in place (from the leaves up to the root)
sort_children_by_leaves(rooted_tree)

fig, ax = plt.subplots(figsize=(10, 10))
plot_nj_tree(rooted_tree, ax=ax, fontsize=9, label_func=short_name)
ax.set_title("c) Rooted tree with children sorted by number of leaves")
ax.set_xlabel("Hamming distance from root (mismatched alignment positions)")
ax.set_yticks([])
ax.spines["left"].set_visible(False)
plt.show()


````

## Problem 4: Recombination

We tend to think of evolution as a hierarchical process: we have some common ancestor that split into several sub-species through a series of random mutations. However, nothing is that straightforward in biology. There have been well-documented cases of horizontal gene transfer in nature where organisms swap entire chunks of DNA with each other. In viruses, something similar can happen as well. When a cell is infected with multiple virions (potentially different strains of the same virus), a *recombination event* can occur. Recombination occurs during the DNA copying process when, for whatever reason, the copier swaps out the virus template midway with a different template. For instance, we could start copying virus A, and then midway, the copier will switch to virus B, effectively resulting in new virus DNA that has DNA from both viruses A and B.

During the start of the global pandemic, there was much confusion about the origins of SARS-CoV-2. We initially thought it came from bats. Then, at some point, people noticed that a critical piece of the spike protein -- the receptor binding domain (RBD) -- was more similar to the sequence of pangolin coronaviruses. This evidence points to recombination. Before the virus jumped to humans, there must have been some host infected with both the Bat RaTG13 coronavirus and the Pangolin coronavirus where this recombination happened.

````python
accession_codes = {
    "Human-SARS-CoV-2": "NC_045512",
    "Bat-CoV RaTG13": "MN996532",
    "Pangolin-CoV MP789": "MT121216",
}


# Step 1: download the genomes, extract the spike proteins and align them

import time
from Bio import SeqIO
from Bio.Align import substitution_matrices
from helper_functions import global_alignment

# Use BLOSUM62 with a linear gap penalty of -4, as in the previous assignment
blosum62 = substitution_matrices.load("BLOSUM62")


def blosum62_score(a, b, gap_penalty=-4):
    if a == "-" or b == "-":
        return gap_penalty
    return blosum62[a, b]

# 1. Download the GenBank records of the three genomes in one request
handle = Entrez.efetch(db="nucleotide", id=",".join(accession_codes.values()), rettype="gb", retmode="text")
records_by_accession = {rec.id.split(".")[0]: rec for rec in SeqIO.parse(handle, "genbank")}
handle.close()
genomes = {name: records_by_accession[acc] for name, acc in accession_codes.items()}

# 2. Extract spike protein sequences from the CDS annotated as gene "S"
def get_spike_protein(record):
    for feature in record.features:
        if feature.type == "CDS" and feature.qualifiers.get("gene", [""])[0] == "S":
            return feature.qualifiers["translation"][0]
    raise ValueError(f"no spike CDS in {record.id}")


spikes = {name: get_spike_protein(record) for name, record in genomes.items()}
for name, protein in spikes.items():
    print(f"{name:20s} genome {len(genomes[name].seq):6d} nt   spike {len(protein):5d} aa")

# 3. Align SARS-CoV-2 spike protein with bat and pangolin spike proteins
# using the Needleman-Wunsch algorithm(from previous assignment)
human = "Human-SARS-CoV-2"
spike_alignments = {}
for other in ["Bat-CoV RaTG13", "Pangolin-CoV MP789"]:
    start = time.time()
    aligned_human, aligned_other, score = global_alignment(spikes[human], spikes[other], blosum62_score)
    spike_alignments[other] = (aligned_human, aligned_other)
    identical = sum(a == b and a != "-" for a, b in zip(aligned_human, aligned_other))
    print(f"spike {human} vs {other}: score {score:.0f}, alignment length {len(aligned_human)}, "
          f"identical {identical / len(spikes[human]):.1%} of SARS-CoV-2 residues ({time.time() - start:.1f} s)")

````

````python
# Step 2: find the RBD and compare similarity inside / outside it with a sliding window

import matplotlib.pyplot as plt

# 1. Identify the RBD position using NCBI protein annotations
spike_cds = next(f for f in genomes[human].features
                 if f.type == "CDS" and f.qualifiers.get("gene", [""])[0] == "S")
handle = Entrez.efetch(db="protein", id=spike_cds.qualifiers["protein_id"][0], rettype="gp", retmode="text")
spike_record = SeqIO.read(handle, "genbank")
handle.close()
rbd_feature = next(f for f in spike_record.features
                   if f.type == "Region" and "RBD" in f.qualifiers.get("region_name", [""])[0])
rbd_start, rbd_end = int(rbd_feature.location.start) + 1, int(rbd_feature.location.end)  # 1-based, inclusive
print(f"RBD of the SARS-CoV-2 spike: residues {rbd_start}-{rbd_end}")


# 2. Identify matching residues at each SARS-CoV-2 position
# Exclude reference gaps to maintain consistent sequence coordinates
def matches_along_reference(aligned_ref, aligned_other):
    return np.array([a == b for a, b in zip(aligned_ref, aligned_other) if a != "-"], dtype=float)


# 3. Sliding window: Calculate sequence identity using a sliding window with a step size of 1
def sliding_window(matches, window=100):
    similarity = np.convolve(matches, np.ones(window) / window, mode="valid")
    centres = np.arange(len(similarity)) + (window + 1) / 2  # 1-based centre of each window
    return centres, similarity


WINDOW = 100
spike_matches = {other: matches_along_reference(*alignment) for other, alignment in spike_alignments.items()}

in_rbd = np.zeros(len(spikes[human]), dtype=bool)
in_rbd[rbd_start - 1:rbd_end] = True
for other, matches in spike_matches.items():
    print(f"{other:20s} identical inside RBD {matches[in_rbd].mean():.1%}   outside RBD {matches[~in_rbd].mean():.1%}")

# 4. Plot both similarity curves along the spike protein, with the RBD shaded
curve_colors = {"Bat-CoV RaTG13": "blue", "Pangolin-CoV MP789": "red"}


def plot_similarity(ax, matches_by_virus, window, xlabel):
    for other, matches in matches_by_virus.items():
        centres, similarity = sliding_window(matches, window)
        ax.plot(centres, similarity * 100, color=curve_colors[other], lw=1.5, label=f"SARS-CoV-2 vs {other}")
    ax.set_xlabel(xlabel)
    ax.set_ylabel(f"Identity in {window}-position window (%)")
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)


fig, ax = plt.subplots(figsize=(12, 4))
ax.axvspan(rbd_start, rbd_end, color="gray", alpha=0.2, label=f"RBD ({rbd_start}-{rbd_end})")
plot_similarity(ax, spike_matches, WINDOW, "Position in SARS-CoV-2 spike protein (amino acid)")
ax.set_title("Spike protein: amino-acid identity to SARS-CoV-2")
ax.legend(loc="lower right", frameon=False)
plt.show()
````

## Bonus 1: Bootstrap confidence

We performed an alignement, calulated pairwise distances and built a phylogenetic tree. Are we confident in our phylogenetic tree? The answer is, we don't know, because we never estimated the uncertainty of our solution.

In 1985, Joseph Felsenstein wrote an [article](https://pubmed.ncbi.nlm.nih.gov/28561359/) on estimating confidence limits of phylogenetic trees using a non-parametric bootstrap. It stands from the idea of creating bootstrap versions of alignment and building alternative phylogenetic trees. A confidence in a clade is then represented as a proportion of alternative trees that contain the same clade.

To generate a bootstrap version of alignment with length $N$, we sample $N$ alignment positions with replacement. Rearanging sequences accoring to a sequence of sampled indicies give us an alternative alignment. Based on new alignment we calculate the distance and apply neighbor joining algorithm to get an alternative tree.

Here we define a clade equivalence:
*Let $L$ be a set of leaf nodes in a clade from original tree and $L'$ be a set of leaf nodes in a clade from alternative tree. Let $L_1$ and $L_2$ be sets of leaf nodes of both children of a clade, such that $L = L_1 \bigcup L_2$. A clade of an alternative tree is considered the same if and only if $L = L'$ and ( $L_1 \bigcap L_1' = \emptyset\;\land\;L_2 \bigcap L_2'=\emptyset$ or $L_1 \bigcap L_2' = \emptyset\;\land\;L_2 \bigcap L_1'=\emptyset$ ).*
